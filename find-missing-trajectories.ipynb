```python
import geopandas as gpd

points = gpd.read_file("points.gpkg", layer="kappazunder_image_punkte")
points.head()
```

wenn ein punkt einer (gesplitteten) trajektorie auf mapillary ist davon ausgehen dass die ganze trajektorie da ist. 

fehlt der punkt fehlt die ganze trajektorie.

```python
from tqdm import tqdm
import mapillary.interface as mly
import json

eli_creator_id = 855778393453994

mly.set_access_token("MLY|28659377196993340|cc6a33649447e29e44ff81a01c42b2d7")

for traj_id, group in tqdm(points.groupby('split_trajectoryid')):
    point = group.sample(1)
    point.geometry = point.geometry.buffer(0.5)
    point = point.to_crs(epsg=4326)
    west, south, east, north = point.total_bounds
    features = json.loads(mly.images_in_bbox({"west": west, "south": south, "east": east, "north": north}))["features"]
    matches = [
        feature
        for feature in features
        if feature["properties"]["creator_id"] == eli_creator_id
    ]
    
    if len(matches) == 0 or len(matches) % 4 != 0:
        print(f"objectid {point.iloc[0].objectid}, trajectoryid {traj_id}, liefert {len(matches)} features an position:")
        print(f"{features}")
```
